```python
outlier_report = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    outlier_report.append({
        "Feature": column,
        "Outliers": len(outliers),
        "Percentage": round(len(outliers) / len(df) * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_report)

outlier_report
```

```python
prophet_df = df[
    ["latest_news_time", "trend_score"]
].copy()
```

```python
prophet_df.columns = [
    "ds",
    "y"
]
```

```python
prophet_df = prophet_df.dropna(
    subset=["ds"]
)
```

```python
prophet_df.shape
```

```python
prophet_df["ds"] = prophet_df["ds"].dt.date
```

```python
prophet_df["ds"] = pd.to_datetime(
    prophet_df["ds"]
)
```

```python
prophet_df = (
    prophet_df
    .groupby("ds", as_index=False)["y"]
    .mean()
)
```

```python
prophet_df = prophet_df.sort_values(
    by="ds"
)
```
